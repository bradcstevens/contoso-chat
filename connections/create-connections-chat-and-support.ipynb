```python
import os
from pathlib import Path

from promptflow import PFClient
from promptflow.entities import (
    CustomConnection,
)
from dotenv import load_dotenv

load_dotenv()

pf = PFClient()
```

```python
# Create local Azure OpenAI Connection
SUPPORT_KEY= os.environ["CONTOSO_SUPPORT_KEY"]
SUPPORT_ENDPOINT= os.environ["CONTOSO_SUPPORT_ENDPOINT"]
connection = CustomConnection(
    name="support-endpoint",
    configs={
        "endpoint": SUPPORT_ENDPOINT,
    },
    secrets={"key": SUPPORT_KEY},
)

print(f"Creating connection {connection.name}...")
result = pf.connections.create_or_update(connection)
print(result)
```
